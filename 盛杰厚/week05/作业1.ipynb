```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("BAAI/bge-small-zh-v1.5")
```

```python
import  numpy as  np
sentences = [
    "我喜欢机器学习",
    "我喜欢深度学习",
    "我今天心情很不错"
]
embeddings = model.encode(sentences, normalize_embeddings=True)
print(embeddings.shape)
query = "我今天很开心"
query_embeddings = model.encode([query], normalize_embeddings=True)
print(query_embeddings.shape)
similarity = query_embeddings @ embeddings.T
similarity = similarity.flatten()
print(similarity[:3])
top_k = 3
# 获取 Top-K
top_k_indices = np.argsort(similarity)[-top_k:][::-1]  # 降序
top_k_scores = similarity[top_k_indices]# shape: (n_docs,)
top_similarity_texts = np.array(sentences)[top_k_indices]
print(f"top_k_indices: {top_k_indices} top_k_scores: {top_k_scores} top_similarity_texts: {top_similarity_texts}")
```
